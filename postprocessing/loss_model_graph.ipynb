```python
import numpy as np
import matplotlib.pyplot as plt

# Constants
epsilon_0 = 8.85e-12  # Permittivity of free space (F/m)
mu_0 = 4 * np.pi * 1e-7  # Permeability of free space (H/m)

# Example parameters
epsilon_r = 80  # Relative permittivity
sigma = 4  # Conductivity in S/m (for seawater)
f = 9e8  # Frequency in Hz (2.4 GHz as an example)

# Derived parameters
omega = 2 * np.pi * f  # Angular frequency
mu = mu_0  # Assumed for nonmagnetic water
epsilon = epsilon_0 * epsilon_r  # Absolute permittivity

# Compute alpha (attenuation factor in Np/m)
alpha = omega * np.sqrt((mu * epsilon) / 2) * (
    np.sqrt(1 + (sigma / (omega * epsilon))**2) - 1
)

# Distance range
D = np.linspace(1, 5, 5)  # Distance in meters

# Path loss computation
tx_power_level = 1  # Transmit power level in dBm
path_loss_db = tx_power_level - alpha * (20 / np.log(10)) * D

# Plot
plt.figure(figsize=(8, 6))
plt.plot(D, path_loss_db, label="Path Loss (dB)", color='b')
plt.xlabel("Distance (m)")
plt.ylabel("Path Loss (dB)")
plt.title(f"Underwater RF Path Loss vs Distance\nepislon_r = {epsilon_r}, sigma = {sigma} S/m, f = {f} Hz")
plt.grid(True)
plt.legend()
plt.show()
```
